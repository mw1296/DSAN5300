### 3.7

$$
R^2 = \frac{\text{SSR}}{\text{SST}} = \frac{\sum_i (\hat{y}_i - \bar{y})^2}{\sum_i (y_i - \bar{y})^2}
$$

Assuming x_bar = y_bar = 0, the slope of the regression line is:

$$
\hat{\beta}_1 = \frac{\sum_i x_i y_i}{\sum_i x_i^2}
$$

The fitted values are:

$$
\hat{y}_i = \hat{\beta}_1 x_i
$$

Then the regression sum of squares (SSR) is:

$$
\text{SSR} = \sum_i (\hat{y}_i - \bar{y})^2 = \sum_i (\hat{\beta}_1 x_i)^2 = \hat{\beta}_1^2 \sum_i x_i^2
$$


$$
\text{SSR} = \left(\frac{\sum_i x_i y_i}{\sum_i x_i^2}\right)^2 \sum_i x_i^2 = \frac{\left(\sum_i x_i y_i\right)^2}{\sum_i x_i^2}
$$

The SST is:

$$
\text{SST} = \sum_i (y_i - \bar{y})^2 = \sum_i y_i^2
$$

So \( R^2 \) is:

$$
R^2 = \frac{\text{SSR}}{\text{SST}} = \frac{\left(\sum_i x_i y_i\right)^2 / \sum_i x_i^2}{\sum_i y_i^2} = \frac{\left(\sum_i x_i y_i\right)^2}{\sum_i x_i^2 \sum_i y_i^2} = \text{Corr}(X,Y)^2
$$



$$
R^2 = \text{Corr}(Y, \hat{Y})^2
$$

$$
\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X
$$
$$
\bar{Y} = \frac{1}{n} \sum_{i=1}^n Y_i
$$
$$
R^2 = 1 - \frac{\text{RSS}}{\text{TSS}}
$$
$$
\text{RSS} = \sum_{i=1}^n (Y_i - \hat{Y}_i)^2, \quad
\text{TSS} = \sum_{i=1}^n (Y_i - \bar{Y})^2
$$


$$
\sum_{i=1}^n (Y_i - \hat{Y}_i) = 0
$$

$$
\sum_{i=1}^n (\hat{Y}_i - \bar{Y})^2 = \frac{(\sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y}))^2}{\sum_{i=1}^n (X_i - \bar{X})^2}
$$

$$
R^2 = \frac{\text{Explained Sum of Squares}}{\text{Total Sum of Squares}}
= \frac{\sum_{i=1}^n (\hat{Y}_i - \bar{Y})^2}{\sum_{i=1}^n (Y_i - \bar{Y})^2}
$$

$$
R^2 = \frac{\left( \sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y}) \right)^2}{\sum_{i=1}^n (X_i - \bar{X})^2 \sum_{i=1}^n (Y_i - \bar{Y})^2}
$$

$$
\text{Corr}(X, Y) = \frac{\sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum_{i=1}^n (X_i - \bar{X})^2} \sqrt{\sum_{i=1}^n (Y_i - \bar{Y})^2}}
$$

$$
\text{Corr}(X, Y)^2 = \frac{\left( \sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y}) \right)^2}{\sum_{i=1}^n (X_i - \bar{X})^2 \sum_{i=1}^n (Y_i - \bar{Y})^2} = R^2
$$

### 3.9 (a)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load Auto dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"
columns = ["mpg","cylinders","displacement","horsepower",
           "weight","acceleration","year","origin","name"]

auto = pd.read_csv(url, delim_whitespace=True, names=columns, na_values="?")
auto.head()
```

```python
# create scatterplot on all variables
sns.pairplot(auto)
```

### 3.9 (b)

```python
# compute matrix of correlations of all variables
df = auto.drop(columns='name')
df.corr()
```

### 3.9 (c)

```python
import statsmodels.api as sm
import numpy as np
```

```python
# multiple linear regression
df = df.dropna()
x = df.drop(columns=['mpg'])
x = sm.add_constant(x) # add constant
y = df['mpg']
mlr_model = sm.OLS(y, x).fit() # fit the model

# summarize model
mlr_model.summary()
```

### 3.9 (c)

```python
from statsmodels.stats.anova import anova_lm
import statsmodels.formula.api as smf
```

```python
# rebuild the model using smf
mlr_model = smf.ols(
    'mpg ~ cylinders + displacement + horsepower + weight + acceleration + year + origin',
    data=df
).fit()

# anova_lm
anova_lm(mlr_model)
```

```python
mlr_model.summary()
```

There is a statistically significant relationship between the most predictors and the response mpg. Only acceleration shows insignificant relationship. The coefficient for the year variable (0.75) suggests a car model of more recent years are built with higher mpg. With each model year advanced, the mpg increases by 0.75 assuming holiding other variables the same.

### 3.9 (d)

```python
import matplotlib.pyplot as plt
```

```python
# plot y pred vs. y

# generate predicted values
y_pred = mlr_model.fittedvalues

plt.scatter(y, y_pred)
# add a diagonal line indicating perfect prediction
line = [y.min(), y.max()]
plt.plot(line, line, '--', color='red')  # red dashed line
plt.xlabel('True mpg')
plt.ylabel('Predicted mpg')
plt.title('True mpg vs. Predicted mpg')
```

```python
# plot residual plot
residuals = mlr_model.resid
plt.scatter(y_pred, residuals)
plt.plot([0,max(y_pred)],[0,0], '--', color='red')
plt.xlabel('Predicted mpg')
plt.ylabel('Residuals')
plt.title('Residuals vs. Predicted mpg')
```

```python
from statsmodels.graphics.regressionplots import plot_leverage_resid2
```

```python
# leverage plot
fig = plot_leverage_resid2(mlr_model)
```

From the leverage plot, along with residual plot and the fitted plot, most observations have low leverage and small residuals (clustered near bottom left 0,0), suggesting most predictors and responses are not outliers. Observation 13, 28, 394 has high leverage, which could influence the predicted reesult. Points on the right side of the graph such as 322, 326, and 325, have larger residuals, which are outliers but not extreme. 

### 3.9 (e)

```python
# fit a model to add interaction of cycliner * displacement, horsepower * weight and year * acceleration
mlr_model_interactions = smf.ols(
    'mpg ~ cylinders + displacement + horsepower + weight + acceleration + year + origin + displacement * cylinders + horsepower * weight + year * horsepower + weight * acceleration',
    data=df
).fit()

mlr_model_interactions.summary()
```

The model with interactions shows that there is small positive but significant interactive effect (coef. = 3.048*10-5) between horsepower and weight, suggesting that the effect of horsepower is dependent on the weight. The horsepower increases MPG more for heavier cars.

### 3.9 (f)

```python
# fit log model
mlr_model_log = smf.ols(
    'mpg ~ cylinders + np.log(displacement) + np.log(horsepower) + np.log(weight) + np.log(acceleration) + year + origin',
    data=df
).fit()

mlr_model_log.summary()
```

```python
# fit square root model
mlr_model_sqrt = smf.ols(
    'mpg ~ cylinders + np.sqrt(displacement) + np.sqrt(horsepower) + np.sqrt(weight) + np.sqrt(acceleration) + year + origin',
    data=df
).fit()

print(mlr_model_sqrt.summary())
```

```python
# fit quadratic model
mlr_model_quad = smf.ols(
    'mpg ~ cylinders + I(displacement**2) + I(horsepower**2) + I(weight**2) + I(acceleration**2) + year + origin',
    data=auto
).fit()

print(mlr_model_quad.summary())
```

```python
# create a table to compare the results

# log results
model_log_results = pd.DataFrame({
    "Coefficient": mlr_model_log.params,
    "P-value": mlr_model_log.pvalues,
    'Adj R Squared': mlr_model_log.rsquared_adj
})

model_log_results = pd.DataFrame({
    "Coefficient": mlr_model_log.params,
    "P-value": mlr_model_log.pvalues,
    'Adj R Squared': mlr_model_log.rsquared_adj
})

print ('Log(X) model result is:')
print(model_log_results.loc[[
    'np.log(horsepower)',
    'np.log(weight)',
    'np.log(acceleration)'
]])

# sqrt(X) results
model_sqrt_results = pd.DataFrame({
    "Coefficient": mlr_model_sqrt.params,
    "P-value": mlr_model_sqrt.pvalues,
    'Adj R Squared': mlr_model_sqrt.rsquared_adj
})

model_sqrt_results = pd.DataFrame({
    "Coefficient": mlr_model_sqrt.params,
    "P-value": mlr_model_sqrt.pvalues,
    'Adj R Squared': mlr_model_sqrt.rsquared_adj
})

print ('\n sqrt(X) model result is:')
print(model_sqrt_results.loc[[
    'np.sqrt(weight)'
]])


# X^2 results
model_quad_results = pd.DataFrame({
    "Coefficient": mlr_model_quad.params,
    "P-value": mlr_model_quad.pvalues,
    'Adj R Squared': mlr_model_quad.rsquared_adj
})

model_quad_results = pd.DataFrame({
    "Coefficient": mlr_model_quad.params,
    "P-value": mlr_model_quad.pvalues,
    'Adj R Squared': mlr_model_quad.rsquared_adj
})

print ('\n quad(X) model result is:')
print(model_quad_results.loc[[
    'I(displacement ** 2)',
    'I(weight ** 2)'
]])
```

Among the three transformations models, the log(X) model has the best fit with the highest adjusted R^2 of 0.847. Three log-transformed predictors are statistically significant - horsepower, weight and acceleration, suggesting nonlinear relationships between MPG and these 3 variables. The square root model performs slighty worse than adjusted R^2 as of 0.831 with one significant tranformed variable (weight^2), while the quadratic model performs worst (adjusted R^2 = 0.799).

```python
# load Weekly data
from ISLP import load_data

Weekly = load_data("Weekly")
Weekly.head()
```

```python
# EDA of the data

# describe data
display(Weekly.describe())

# aggregate data by year
year_agg = Weekly.groupby('Year').agg(
  volume_mean = ('Volume', 'mean'),
  volume_min = ('Volume', 'min'),
  volume_max = ('Volume', 'max'),
  today_mean = ('Today','mean'),
  today_min = ('Today', 'min'),
  today_max = ('Today', 'max'),
  pct_up_weeks = ('Direction', lambda x: (x=='Up').mean())
)

year_agg = year_agg.reset_index()
year_agg
```

We can observe below pattens from the graph:

2. Huge surge of trading volume from 2006 till 2009. 2010 dropped back to 2007 level.

### (b)

```python
# logistic regression

import statsmodels.api as sm
Weekly['direction_encode'] = Weekly['Direction'].map({'Up': 1, 'Down': 0 }) # transform 'up' as 1 and 'down' as 0
x = Weekly[['Lag1','Lag2','Lag3','Lag4','Lag5','Volume']]
x = sm.add_constant(x) # add constant
y = Weekly['direction_encode']

# fit logit model
logit_model = sm.Logit(y, x)
logit_model_result = logit_model.fit()
logit_model_result.summary()
```

The logistic regression result shows that only Lag2 is statistically significant (p = 0.03). All other predictors have p-values greater than 0.05 and are not significant. The pseudo R-squared is very low (0.0066), suggesting that the model explains very little of the variation in weekly market direction.

### (c)

```python
# make predictions
pred_prob = logit_model_result.predict(x)

# convert >=0.5 as "up" and <0.5 as "down"
Weekly['direction_pred'] = ["Up" if x>=0.5 else 'Down' for x in pred_prob]
```

```python
# plot confusion matrix
from sklearn.metrics import confusion_matrix, accuracy_score, ConfusionMatrixDisplay

disp_frac = ConfusionMatrixDisplay.from_predictions(
    Weekly['Direction'],
    Weekly['direction_pred'],
    normalize="true"
)

plt.title("Confusion Matrix (fraction)")
plt.show();

disp = ConfusionMatrixDisplay.from_predictions(
    Weekly['Direction'],
    Weekly['direction_pred']
)
plt.title("Confusion Matrix")
plt.show();
```

The confusion matrix shows the logistic model predicts majority of the observations as "Up" (90%). Although the Recall rate on "Up" is high (92%), the Precision (56%) on "Up" is low. Both the Recall rate (11%) and Precision (54%) on "Down" is low accordingly. Overall, due to the biases prediction towards "Up", the model shows low accuracy score.
