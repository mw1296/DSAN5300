**Batch (full-batch) gradient descent**

* Uses the *entire training dataset* to compute each gradient update
* Deterministic, stable convergence
* Computationally expensive per step; slow for large (N)

**Mini-batch gradient descent**

* Uses a *subset* of size ($B \ll N$) per update
* Tradeoff: efficiency + moderate gradient noise
* Standard in practice (GPU-friendly, faster convergence)

**Stochastic gradient descent (SGD)**

* Uses *one sample* in the training dataset per update ((B=1))
* Very noisy gradients; high variance
* Fast updates, can escape shallow local minima, but less stable
